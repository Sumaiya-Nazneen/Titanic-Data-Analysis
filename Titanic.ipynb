```python
ticket_pass=df['ticket'].value_counts().reset_index(name='no_of_psgs')
```

```python
ticket_pass
df
```

```python
df=pd.merge(df,ticket_pass,on='ticket',how='inner')
```

```python
df
```

```python
df['fare']=(df['fare']/df['no_of_psgs'].round(1))
```

```python
df['fare']
```

```python
def comp_type(row):
    f=row['fam_cnt']
    p=row['no_of_psgs']
    if p==1:
        return 'Solo'
    elif f==0 and p>1:
        return 'friends'
    else:
        return 'family'
```

```python
df['Companion Type']=df.apply(comp_type,axis=1)
```

```python
df['Age_category']=pd.cut(df['age'],bins=[0,2,16,25,59,100],labels=['infant','child','teen','adult','senior'])
df
```

```python
df['Title'].unique()
```

```python
Royal=['the Coum=ntless','Jonkheer','Don','Dona','Master','Sir']
Common=['Miss','Mr','Lady','Ms','Mme','Mile']
Military=['Major','Col','Capt']
other_profession=['Dr','Rev']
```

```python
def social_status(row):
    c=row['Title']
    if c in Royal:
        return 'Royal'
    elif c in Common:
        return 'Common'
    elif c in Military:
        return 'Military'
    else:
        return other_profession
df['Social_status']=df.apply(social_status,axis=1)
```

```python
plt.figure(figsize=(3,4))
cnt=sns.countplot(data=df,x='pclass',color='olive',order=[3,1,2])
cnt.bar_label(cnt.containers[0])
plt.title("NO OF PASSENGERS IN EACH CLASS")
plt.yticks(range(0,900,100))
plt.ylabel("NO OF PASSENGERS")
plt.show()
```

**OBSERVATION**



* Most passengers are adults, and the number of male passengers is higher than female passengers.
